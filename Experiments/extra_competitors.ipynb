```python
dataset_name, df = read_magic(False)#read_compas()

#to_encode = ['MonthlyCharges', 'TotalCharges', 'tenure']  # magic
to_encode = df.columns[:-1]

for col in to_encode:
    df[col] = pd.qcut(df[col], 10)

for col in df.columns:
    df[col] = df[col].apply(lambda x: f'{x}|{col}')

df.head()
```

```python
from sklearn.model_selection import train_test_split

X = df#[['y']+list(df.columns)[:-1]].values.tolist()
y = df.y#.values.tolist()

X_train_df, X_test_df, y_train_df, y_test_df = train_test_split(X, y, stratify=y, random_state=42, test_size=.2)

X_train, X_test, y_train, y_test = (X_train_df.values.tolist(), X_test_df.values.tolist(), 
                                    y_train_df.values.tolist(), y_test_df.values.tolist())

X_train[0]
```

```python
from fim import apriori
help(apriori)
```

```python
supp = 5
zmin = 2

rules = apriori(X_train, target="r", supp=supp, zmin=zmin, conf=80, report="aScl")
rules_df = pd.DataFrame(rules, columns=["consequent", "antecedent", "abs_support", "%_support", "confidence", "lift"])

rules_df.sort_values(by="lift", axis=0, ascending=False)
```

```python
rules_df[rules_df.consequent.apply(lambda x: f'{x}'.endswith('|y'))]
```

```python
def codifica(df_rules, df):
    rules = df_rules.antecedent.unique()
    X_cod = np.ones((len(df), len(rules))).astype(np.bool_)
    for i, rule_list in enumerate(rules):
        for rule in rule_list:
            feat = rule.split('|')[-1]
            X_cod[:, i] &= (df[feat] == rule).values
    return X_cod, rules
```

```python
rules_df_cod = rules_df[rules_df.antecedent.apply(lambda x: '|y' not in str(x)) & rules_df.consequent.apply(lambda x: '|y' not in x)]

X_train_cod, rules = codifica(df_rules=rules_df_cod, df=X_train_df)
X_test_cod, _ = codifica(df_rules=rules_df_cod, df=X_test_df)
```

```python
rules_df_cod
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report

le = LabelEncoder()
y_train_num = le.fit_transform(y_train)
y_test_num = le.transform(y_test)


lr = LogisticRegression()
lr.fit(X_train_cod, y_train_num)

y_pred = lr.predict(X_test_cod)

print(classification_report(y_true=y_test_num, y_pred=y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier, export_text

rt = DecisionTreeClassifier(max_depth=1)
rt.fit(X_train_cod, y_train_num)

y_pred = rt.predict(X_test_cod)

print(classification_report(y_true=y_test_num, y_pred=y_pred))
```

```python
print(export_text(rt, feature_names=rules))
```
