```python
from Experiments.readers import all_datasets
import pandas as pd
import numpy as np
```

```python
datasets = dict()

for d in all_datasets:
    k, v = d()
    n_rows, n_feat = v.shape
    n_cat = np.sum([len(v[feat].unique()) == 2 for feat in v.columns])
    M, m = max(np.unique(v.y, return_counts=True)[1]), min(np.unique(v.y, return_counts=True)[1])
    datasets[k] = {
        'n_rows': n_rows,
        'n_feat': n_feat,
        'n_cat': n_cat,
        'n_cont': n_feat-n_cat,
        'max': round(M/n_rows, 3)*100,
        'min': round(m/n_rows, 3)*100
    }
```

```python
for k in sorted(datasets):
    print(k, datasets[k])
```

```python
print(pd.DataFrame(datasets)[sorted(datasets)].to_latex())
```

```python
pd.DataFrame(datasets).to_csv('metadata_dataset.csv')
```
